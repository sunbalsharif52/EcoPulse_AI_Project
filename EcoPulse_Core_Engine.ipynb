```python
import yfinance as yf
import pandas as pd
from datetime import datetime, timedelta

print("Fetching Brent crude oil data from Yahoo Finance...")

# Calculate the date range for the last three years
end_date = datetime.today().strftime("%Y-%m-%d")
start_date = (datetime.today() - timedelta(days=3 * 365)).strftime("%Y-%m-%d")

# Download Brent crude oil futures data
ticker = "BZ=F"
oil_data = yf.download(
    ticker,
    start=start_date,
    end=end_date,
    progress=False
)

# Check and process the downloaded data
if not oil_data.empty:
    if isinstance(oil_data.columns, pd.MultiIndex):
        oil_data.columns = oil_data.columns.get_level_values(0)

    oil_df = oil_data[["Close", "Volume"]].copy()

    print(f"Data downloaded successfully. Total records: {len(oil_df)}")
    print("\nData Preview (First Five Rows):")
    print(oil_df.head())
else:
    print("Data download failed. Check your internet connection or ticker availability.")
```
